```python
import json
```

```python
res_files = {
    "constraint_1": "constraint_2024_07_19_16_56_38",
    "constraint_2": "constraint_2024_07_19_16_34_48",
    "weblinx": "weblinx_baseline_2024_07_19_16_46_14",
}

results = {}
for k, v in res_files.items():
    file_name = "../results/" + v + "/results.json"
    with open(file_name, "r") as f:
        r = json.load(f)
    results[k] = r
```

```python
import matplotlib.pyplot as plt

for label, res in results.items():
    recall_dict = {k: v for k, v in res.items() if 'recall' in k}
    recall_vals = [k.split('@')[-1] for k in recall_dict.keys()]
    plt.plot(recall_vals, list(recall_dict.values()), label=label)
plt.legend()
plt.ylim(0,1)
```
