```python
import json
import matplotlib.pyplot as plt
```

```python
def plot_res_files(res_files):
    results = {}
    for k, v in res_files.items():
        file_name = "../results/" + v + "/results.json"
        with open(file_name, "r") as f:
            r = json.load(f)
        results[k] = r

    for label, res in results.items():
        recall_dict = {k: v for k, v in res.items() if 'recall' in k}
        recall_vals = [k.split('@')[-1] for k in recall_dict.keys()]
        plt.plot(recall_vals, list(recall_dict.values()), label=label)
    plt.legend()
    plt.ylim(0,1)
    plt.xlabel('@')
    plt.ylabel('Recall')

    return results
```

```python
# including invalid uids
res_files = {
    "weblinx": "weblinx_baseline_2024_07_19_16_46_14",
    "constraint": "constraint_2024_07_19_16_56_38",
    # "constraint_2": "constraint_2024_07_19_16_34_48",
    # "webtree_1": "webtree_2024_07_29_11_12_16",
    "webtree_100": "webtree_2024_07_29_11_15_07",
    "webtree_90": "webtree_2024_07_29_11_20_28",
}
_ = plot_res_files(res_files)
```

```python
# only valid uids
res_files = {
    "weblinx": "weblinx_baseline_2024_07_29_11_46_02",
    "constraint": "constraint_2024_07_29_11_42_50",
    "webtree_100": "webtree_2024_07_29_11_33_51",
    "webtree_90": "webtree_2024_07_29_11_28_30",
}
results = plot_res_files(res_files)
```
